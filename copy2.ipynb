```python
import copy

# ==================== 浅いコピー vs 深いコピー ====================

print("=" * 60)
print("1. リストの場合")
print("=" * 60)

# 元のリスト（ネストされたリスト）
original_list = [1, 2, [3, 4, 5]]

# 浅いコピー
shallow_copy_list = copy.copy(original_list)

# 深いコピー
deep_copy_list = copy.deepcopy(original_list)

print(f"元のリスト: {original_list}")
print(f"浅いコピー: {shallow_copy_list}")
print(f"深いコピー: {deep_copy_list}")

# ネストされたリストを修正
original_list[2][0] = 999

print(f"\n元のリスト[2][0]を999に変更後:")
print(f"元のリスト: {original_list}")
print(f"浅いコピー: {shallow_copy_list}  ← 変更される（同じオブジェクト参照）")
print(f"深いコピー: {deep_copy_list}  ← 変更されない（独立したオブジェクト）")

print("\n" + "=" * 60)
print("2. 辞書の場合")
print("=" * 60)

# 元の辞書（ネストされた辞書）
original_dict = {"name": "Alice", "info": {"age": 30, "city": "Tokyo"}}

# 浅いコピー
shallow_copy_dict = copy.copy(original_dict)

# 深いコピー
deep_copy_dict = copy.deepcopy(original_dict)

print(f"元の辞書: {original_dict}")
print(f"浅いコピー: {shallow_copy_dict}")
print(f"深いコピー: {deep_copy_dict}")

# ネストされた辞書を修正
original_dict["info"]["age"] = 35

print(f"\n元の辞書['info']['age']を35に変更後:")
print(f"元の辞書: {original_dict}")
print(f"浅いコピー: {shallow_copy_dict}  ← 変更される")
print(f"深いコピー: {deep_copy_dict}  ← 変更されない")

print("\n" + "=" * 60)
print("3. オブジェクトID比較")
print("=" * 60)

original = [1, 2, [3, 4]]
shallow = copy.copy(original)
deep = copy.deepcopy(original)

print(f"元のリストのID: {id(original)}")
print(f"浅いコピーのID: {id(shallow)}")
print(f"深いコピーのID: {id(deep)}")
print(f"\n浅いコピーの内部リストのID: {id(shallow[2])}")
print(f"深いコピーの内部リストのID: {id(deep[2])}")
print(f"元のリストの内部リストのID: {id(original[2])}")
print(f"\n浅いコピーと元の内部リストは同じ? {shallow[2] is original[2]}")
print(f"深いコピーと元の内部リストは同じ? {deep[2] is original[2]}")
```
