```python
import copy

def show_copy_behavior():
    # 元データ: 1階層目も2階層目も可変
    original = {
        "name": "Alice",
        "scores": [90, 85],
        "meta": {"age": 20, "city": "Tokyo"},
    }

    print("=== 1. 代入（コピーではない） ===")
    alias = original
    alias["scores"].append(100)
    alias["meta"]["city"] = "Kyoto"
    print("original:", original)
    print("alias is original:", alias is original)

    print("\n=== 2. 浅いコピー ===")
    original = {
        "name": "Alice",
        "scores": [90, 85],
        "meta": {"age": 20, "city": "Tokyo"},
    }
    shallow = copy.copy(original)
    shallow["scores"].append(88)
    shallow["meta"]["city"] = "Osaka"

    print("same top-level object?", shallow is original)
    print("same nested list object?", shallow["scores"] is original["scores"])
    print("same nested dict object?", shallow["meta"] is original["meta"])
    print("original:", original)
    print("shallow:", shallow)

    print("\n=== 3. 深いコピー ===")
    original = {
        "name": "Alice",
        "scores": [90, 85],
        "meta": {"age": 20, "city": "Tokyo"},
    }
    deep = copy.deepcopy(original)
    deep["scores"].append(99)
    deep["meta"]["city"] = "Fukuoka"

    print("same top-level object?", deep is original)
    print("same nested list object?", deep["scores"] is original["scores"])
    print("same nested dict object?", deep["meta"] is original["meta"])
    print("original:", original)
    print("deep:", deep)

show_copy_behavior()
```
