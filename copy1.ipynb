```python
# 浅いコピーの例（Jupyter の新しいセル用）
import copy

orig = [[1, 2], [3, 4]]
shallow = copy.copy(orig)   # 浅いコピー（最上位のリストのみコピー）
deep = copy.deepcopy(orig)  # 深いコピー（ネストされたオブジェクトもコピー）

print("変更前")
print("orig   :", orig)
print("shallow:", shallow)
print("deep   :", deep)

# ネスト内の要素を変更（浅いコピーでは共有される）
orig[0][0] = 99

print("\norig[0][0] を 99 に変更した後")
print("orig   :", orig)
print("shallow:", shallow)  # 内部リストは同じ参照なので影響を受ける
print("deep   :", deep)     # 深いコピーは影響を受けない

# 同一性の確認
print("\n同一性チェック（id）")
print("orig is shallow:", orig is shallow)
print("orig[0] is shallow[0]:", orig[0] is shallow[0])
print("orig[0] is deep[0]:", orig[0] is deep[0])
```
